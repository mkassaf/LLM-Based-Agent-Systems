```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
import matplotlib.pyplot as plt
import numpy as np
```

```python
manual_df = pd.read_csv("data/clean_sample_agent_repos_withShortReadme_manual.csv")
model_df = pd.read_csv("data/sample_agent_repos_llm_filtered_withFullReadme_DeepSeekR1_final.csv")
```

```python
merged_df = pd.merge(
    manual_df,
    model_df[["full_name", "category"]],
    on="full_name",
    how="inner" # keeps only rows with matching full_name in both DataFrames
)
```

```python
print("Manual size:", len(manual_df))
print("Model size:", len(model_df))
print("Merged size:", len(merged_df))
```

```python
merged_df.head(3)
```

```python
y_true = merged_df["manual_normalized"]
y_pred = merged_df["category"]
```

```python
accuracy = accuracy_score(y_true, y_pred)

precision = precision_score(y_true, y_pred, average="macro", zero_division=0)
recall = recall_score(y_true, y_pred, average="macro", zero_division=0)
f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1:", f1)

print("\nDetailed Report:\n")
print(classification_report(y_true, y_pred))
```

```python
precision_1 = precision_score(y_true, y_pred, pos_label="llm-based agentic system")
recall_1 = recall_score(y_true, y_pred, pos_label="llm-based agentic system")
f1_1 = f1_score(y_true, y_pred, pos_label="llm-based agentic system")

print("Precision:", precision_1)
print("Recall:", recall)
print("F1:", f1)
```

```python
# Compare
merged_df["Correct"] = (
    merged_df["Manual"] == merged_df["category"]
)

# Save
merged_df.to_csv("comparison_results_fullReadme.csv", index=False)

print("Saved comparison file!")
```
